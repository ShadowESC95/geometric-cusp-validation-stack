## 4. Established \(z\)-contour benchmark

The benchmark finite functions are

\[F^{(0)}(z)=-1,\qquad F^{(1)}(z)=\log^2z+\pi^2.\]

The source convention relates the contour functional to the cusp anomalous dimension. Keep the sign/orientation convention explicit when comparing a new geometric period.

```python
from geocusp.contours import contour_average, benchmark_F0, benchmark_F1, gamma_weak
for f in (benchmark_F0, benchmark_F1):
    avg,_,_=contour_average(f, n=4096)
    print(avg)
print('weak cusp at g=0.1:', gamma_weak(0.1))
```
